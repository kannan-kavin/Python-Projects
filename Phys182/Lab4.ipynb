```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import pi,sqrt,log,exp,sin,cos
import scipy
from scipy.optimize import curve_fit
import pandas as pd
```

```python
trial25 = pd.read_csv("25.csv")
t1 = trial25["Time (s) Run #15"].to_numpy()
V1 = trial25["Voltage, Ch A (V) Run #15"].to_numpy()
t1 = t1[15:36]
t1 = t1 - t1[0]
V1 = V1[15:36]
ln_V1 = np.log(V1)
plt.scatter(t1,ln_V1)
plt.title('Natural Log of Voltage vs Time, R = 25 kΩ')
plt.xlabel('Time [s]')
plt.ylabel('Natural Log of Voltage [ln(V)]')
plt.grid()
plt.show()
```
