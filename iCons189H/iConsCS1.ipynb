```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import pi,sqrt,log,exp,sin,cos
import scipy
from scipy.optimize import curve_fit
```

```python
data = np.loadtxt("cs1.txt")

sct = data[:, 4]
caf = data[:, 3]
gam = data[:, 2]
soc = data[:, 1]
ent = data[:, 0]
sigma = np.zeros(36)
sigma2 = np.sqrt(caf)

plt.scatter(sct, caf)
plt.xlabel("Phone Screen Time (h)")
plt.ylabel("Caffeine Consumption (mg)")
plt.title("UMass Amherst Students Caffeine vs Screen Time over Feb 16-22, 2025")
```

```python
def linear(x, a0, a1):
    return a0 + a1 * x

popt_lin, pcov_lin = curve_fit(linear, caf, sct)

perr_lin = np.sqrt(np.diag(pcov_lin))

x_fit = np.linspace(min(caf), max(caf), 36)

y_fit_lin = linear(x_fit, *popt_lin)

plt.errorbar(caf, sct, yerr=sigma, xerr=sigma2, fmt='o', label='Data', capsize=3)
plt.plot(x_fit, y_fit_lin, label=f'Line of Best Fit')
plt.grid()
plt.xlabel('Total Caffiene from Drinks during the Week (mg)')
plt.ylabel('Total Screen Time from Phones during the Week (h)')
plt.legend()
plt.title('UMass Amherst Students\' Caffiene Intake vs Screen Time over Feb 16-Feb 22, 2025')
plt.show()

# Compute Unweighted Chi-Squared
y_fit_values = linear(caf, *popt_lin)
chi_squared = np.sum((sct - y_fit_values) ** 2)

# Compute Reduced Chi-Squared (assuming 2 parameters)
reduced_chi_squared = chi_squared / (len(caf) - 2)

print(f"{chi_squared}")
print(f"{reduced_chi_squared}")

z, cov = np.polyfit(caf, sct, 1, cov=True)
delta_1 = np.sqrt(cov[0,0])
delta_2 = np.sqrt(cov[1,1])
print('The slope is ' + "{:.4}".format(z[0]) + " +/- " + "{:.2}".format(delta_1))
print('The intercept is ' + "{:.5}".format(z[1]) + " +/- " + "{:.2}".format(delta_2))
```

```python
popt_lin, pcov_lin = curve_fit(linear, caf, ent)

perr_lin = np.sqrt(np.diag(pcov_lin))

x_fit = np.linspace(min(caf), max(caf), 36)

y_fit_lin = linear(x_fit, *popt_lin)
dof = len(caf) - (1 + 1)
residuals = (sct - y_fit_lin) / sigma
chi2 = np.sum(residuals**2)
redchi2lin = chi2/dof

plt.errorbar(caf, ent, yerr=sigma, xerr=sigma2, fmt='o', label='Data', capsize=3)
plt.plot(x_fit, y_fit_lin, label=f'Line of Best Fit')
plt.grid()
plt.xlabel('Total Caffiene in the Week (mg)')
plt.ylabel('Total Screen Time in the Week (h)')
plt.legend()
plt.title('Entertainment')
plt.show()

z, cov = np.polyfit(caf, ent, 1, cov=True)
delta_1 = np.sqrt(cov[0,0])
delta_2 = np.sqrt(cov[1,1])
print('The slope is ' + "{:.4}".format(z[0]) + " +/- " + "{:.2}".format(delta_1))
print('The intercept is ' + "{:.5}".format(z[1]) + " +/- " + "{:.2}".format(delta_2))
```
