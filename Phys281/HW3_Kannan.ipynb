Part 1: Solution to the Lorenz equations in the non-chaotic regime
Consider the Lorenz equations for the case

σ = 10 , ρ = 13 ,β = 8/3 .

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import scipy
from scipy.integrate import odeint
from mpl_toolkits.mplot3d import Axes3D
```

A. (20 pts) Write a program that solves the Lorenz equations for these values of the parameters in
the range t = 0 to t = 50 with initial conditions (x, y, z) = (0,1,0). Keep in mind that these are
coupled equations. Define a function that calculates the three derivatives much as we have done
for other coupled differential equations in class. Be sure to include comments.

```python
def Lorenz(initialconditions, t, sigma, rho, beta): #Creating the function that will compute the 3 time derivatives for x, y, and z according to the Lorenz equations
    x,y,z = initialconditions #Setting the intial conditions x, y, and z
    sigma, rho, beta = sigma, rho, beta #Setting the parameters sigma, rho, and beta for the equations below
    dxdt = sigma * (y - x) #Using the given Lorenz equations to find their time derivatives
    dydt = rho * x - y - x * z
    dzdt = x * y - beta * z
    return[dxdt, dydt, dzdt] #Returns the couples derivatives as defined by the Lorenz equations
```

B. (5 pts) Have your program make a plot of y(t) and z(t) as a function of time on the same graph.
Include a title, legend, and axis labels.

```python
t = np.linspace(0, 50.01, 5000) #Creating the time interval from 0 to 50 seconds
solution = odeint(Lorenz, [0,1,0], t, args=(10, 13, 8/3)) #Using odeint to solve the Lorenz equations with given initial conditions
x,y,z = solution.T #Transposing the returned vector from the Lorenz function to associate each variable to its corresponding integrated derivative

plt.plot(t, y, label='y(t)', color='r') #Plotting y and z vs t on the same graph.
plt.plot(t, z, label='z(t)', color='b') 
plt.xlabel('Time (s)')
plt.ylabel('Value (unitless)')
plt.legend()
plt.grid()
plt.title('Lorenz Equations: y(t) and z(t) ($\\sigma=10$, $\\rho=13$, $\\beta=8/3$)')
plt.show()
```
