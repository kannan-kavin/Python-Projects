Solve a system of linear equations (10 pts) Use Python to solve
the following system of linear equations:
x + 4y + z = −4,
4x − y − z = 3,
2x + y + 3z = 7.

for the three unknowns x, y and z. Do this two different ways: (a) by
multiplying the inverse of the matrix of coefficients by the RHS vector,
and (b) using linalg.solve (which is more accurate in general), and
check that both methods agree. Plug the solution for x, y, z into one
of the original equations to check that it is satisfied.

```python
m = np.array([[1,4,1],[4,-1,-1],[2,1,3]], float)
k = np.array([-4,3,7], float)
minverse = linalg.inv(m)
x = minverse@k

print(f'x = {x[0]}, y = {x[1]}, z = {x[2]}')
```

```python
x = linalg.solve(m,k)

print(f'x = {x[0]}, y = {x[1]}, z = {x[2]}')
```

```python
ss = np.array([[0,np.sqrt(3)/2,0,0], [np.sqrt(3)/2,0,1,0], [0,1,0,np.sqrt(3)/2], [0,0,np.sqrt(3)/2,0]], float)
print('S =\n',ss)
```

```python
evals, evecs = linalg.eigh(ss)
print(evals)
print('\n',evecs)
```

```python
v1 = evecs[:, 0]
print(ss@v1)
```

This result is a scale factor of -1.5 applied to the first eigenvector which is one of our eigenvalues so we're good to go! :)
