###**8. The conical pendulum**

(a) (5 points) The period of a conical pendulum is $$T = 2\pi \sqrt{\frac{L\cos\beta}{g}},$$ where $\beta$ is the angle in radians between the pendulum string and the vertical. Write a program to calculate and print $T$ from user input of $L$ and $\beta$ **in degrees**. As with the last problem, your program should tell the user what units it wants $L$ and $\beta$ to be in, and give the units for the answer $T$. (Note that the cosine function in numpy expects the input in radians.)

```python
g = 9.8
L = float(input("Enter L in meters:"))
Bdeg = float(input("Enter B in degrees:"))
Brad = Bdeg * pi / 180

T = 2 * pi * sqrt(L * cos(Brad) / g)
print(f"The period of the conical pendulum is {T} seconds")
```

###**9. Newman Exercise 2.2: Altitude of a satellite**

(a) (10 points) A satellite is to be launched into a circular orbit around the Earth so that it orbits the planet once every $T$ seconds. Using Newton's second law $F = ma$ for circular motion, where $F = GMm/r^2$ is the gravitational force on the satellite, derive the expression (by hand, not numerically) for the satellite's altitude $h$ above the Earth’s surface and show that the form is:
$$
h = \left(\frac{GMT^2}{4\pi^2}\right)^{1/3} - R
$$
where $G = 6.67 \times 10^{-11} m^3 kg^{-1} s^{-2}$ is Newton's gravitational constant, $M = 5.97 \times 10^{24}$ kg is the mass of the Earth, and R = 6371 km is its radius.  Show this derivation in a markdown cell below.

Derivation:

$F = ma_c = \frac{GMm}{r^2} = \frac{mv^2}{r} = \frac{m4\pi^2}{T^2r}$ through simple rearrangement and replacement

$\frac{GM}{r} = \frac{4\pi^2r^2}{T^2}$ arises from cancellation between the second and fourth terms in the equation

$r^3 = \frac{GMT^2}{4\pi^2}$ through isolating for r

$r = (\frac{GMT^2}{4\pi^2})^\frac{1}{3}$ is the final equation for the distance from the center of earth to the sattelite

$h = (\frac{GMT^2}{4\pi^2})^\frac{1}{3} - R$ is the final equation for ALTITUTE!


(b) (10 points) Write a program that asks the user to enter the desired value of $T$ and then calculates and prints out the correct altitude in meters.
The program must print the error message `Period not allowed` if the height turns out to be negative.

```python
G = 6.67e-11
M = 5.97e24
R = 6371e3
T = float(input("What is your desired period in seconds?"))
h = (G * M * T**2 / (4*pi**2))**(1/3) - R
if h < 0:
    print("Period not allowed")
else:
    print("The altitude in meters is",h)
```

###**11. Newman Exercise 2.7: Catalan numbers.**  (10 points)  

These are a sequence of integers defined in the following way:
\begin{equation}
C_n =\frac{(2n)!}{(n+1)!n!}
\end{equation}
They also satisfy the recurrence relatiion
\begin{equation}
 C_0 = 1, C_{n+1} =\frac{(4n+2)}{n+2}C_n
\end{equation}
Print in order all the Catalan numbers less than or equal to $10^9$. *Remember you want these to be integers, not floats.* You may use the factorial function you defined in the previous exercise.

Look at the Wikipedia entry [Catalan number](https://en.wikipedia.org/wiki/Catalan_number) which lists the first few Catalan numbers to check if your program is working (and also find out some interesting uses of these numbers).

```python
def factorial(n):
    s = 1
    while (n > 0):
        s = s * n
        n = n - 1
    return(s)
n = 0
C = 0
while C <= 10e9:
    first = 2*n
    second = n + 1
    C = (factorial(first) / (factorial(second) * factorial(n)))
    n += 1
    print(C)
```
