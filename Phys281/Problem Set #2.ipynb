**4. User defined functions: Alternating Harmonic Series**

(a) *(5 points)* Define a function `sum(N)` that calculates the sum of the first N terms in an alternating harmonic series:
$$
\verb7sum(N)7 = \sum_{n=1}^N {(-1)^{n-1} \over n}
$$
where the argument of the function is the upper limit of the sum.

(b) *(5 points)* Check, for large N, how close you get to the expected convergent value of this series: $\ln{2}$ (i.e. calculate the % difference between your value and the expected value).

````python
def sum(N):
    sum = 0
    for n in range (1, N + 1, 1):
        sum += ((-1)**(n-1)) / n 
    return(sum)
````

**5. Geometric Series**

(a) (5 points) Define a function `geosum(x, N)` that evaluates the sum of the first N terms of a geometric sequence:
$$
\text{geosum(x, N)}= \sum_{n=0}^{N-1} x^n
$$
Check that this gives you an answer close to the expected value:
$ {(1 - x^N)}/{(1 - x)}$.

(b) (5 points)  The geometric series will not converge unless $|x| < 1$. Add in a statement that checks for this.

````python
def geosum(x, N):
    sum = 0
    for n in range(0, N, 1):
        sum += x**n
    if(abs(x) >= 1):
        print("The series will not converge")
    return(sum)

print(geosum(0.5, 10))
print((1 - 0.5**10) / (1 - 0.5))
````
