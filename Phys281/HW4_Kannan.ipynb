A. (5 pts) Make a plot of the initial transverse displacement of the string as a function of x, �(�,� =
0), making sure it depicts the correct initial triangular pulse described above.

```python
L = 1.0  #Length of the string (m)
velocity_wave = 0.005  #Wave speed (m/s)
xstep = 0.0025  #Step size (m)
xpositions = np.arange(0, L + xstep, xstep)  #Array of positions along the string
t = 500  #Total simulation time (s)

pulse_center = 0.50  #Pulse centered (m)
pulse_height = 0.01  #Pulse height (m)
pulse_width = 0.08  # Pulse width (m)

x0 = pulse_center - pulse_width / 2 #Initial x position given by the problem as well as numerically
xf = pulse_center + pulse_width / 2 #Final x position given by the problem as well as numerically

ypositions0 = np.zeros_like(xpositions) #Initializing an array for the y positions

for i in range(len(xpositions)): #Iterating through every position on the x-axis for t=0
    x = xpositions[i]
    if x0 <= x <= pulse_center: #Altering the yposition of any x that's part of the triangular initial transverse displacement
        ypositions0[i] = (x - x0) / (pulse_width / 2) * pulse_height #Before the center of the pulse
    elif pulse_center < x <= xf:
        ypositions0[i] = (xf - x) / (pulse_width / 2) * pulse_height #After the center of the pulse

# Plot
plt.plot(xpositions, ypositions0)
plt.title("Initial Transverse Displacement along Plucked String: y(x, 0)")
plt.xlabel("Position along the String (m)")
plt.ylabel("Transverse Displacement of the String (m)")
plt.grid()
plt.show()
```
